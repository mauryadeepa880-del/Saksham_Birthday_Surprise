# Saksham's Birthday Surprise 💗
Run the single code cell below. It serves the website from this folder and tries to open it in your browser.

**Keep the cell running** while using the site. To stop the server, interrupt the cell.


```python
from http.server import ThreadingHTTPServer, SimpleHTTPRequestHandler
from pathlib import Path
import os, threading, webbrowser

os.chdir(Path.cwd())  # Jupyter should be opened in the extracted project folder
print('Project folder:', Path.cwd())
server = ThreadingHTTPServer(('127.0.0.1', 8000), SimpleHTTPRequestHandler)
url = 'http://127.0.0.1:8000'
print('Birthday surprise is running at:', url)
print('If a browser does not open, copy the address above into Chrome.')
threading.Timer(1.0, lambda: webbrowser.open(url)).start()
try:
    server.serve_forever()
except KeyboardInterrupt:
    print('Website stopped.')
    server.server_close()
```
